```python
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

plt.rcParams['figure.figsize']=(11,6)
plt.rcParams['font.size']=11
plt.rcParams['axes.titlesize']=16
print('Analysis environment ready.')
```

## Table 1 — Registration time

The thesis reports 342 observations in each condition.

```python
table1=pd.DataFrame({'Registration time':['2–4 months','1–2 months','1–2 weeks','4–6 days','1–2 days','2–4 hours','1–2 hours','15–30 minutes','2–15 minutes'],'Without QR system':[93,50,120,55,24,0,0,0,0],'With QR system':[0,0,0,0,0,58,120,59,105]})
display(table1)
print('Totals:',table1['Without QR system'].sum(),table1['With QR system'].sum())
```

```python
ax=table1.set_index('Registration time').plot(kind='bar',figsize=(12,7))
ax.set_title('Table 1 — Registration Time: Without QR vs. With QR',fontweight='bold')
ax.set_xlabel('Registration time'); ax.set_ylabel('Number of users'); ax.legend(title='Condition'); ax.grid(axis='y',alpha=.25)
plt.xticks(rotation=25,ha='right'); plt.tight_layout(); plt.show()
```

**Interpretation:** The largest group changes from **1–2 weeks (120 users)** without QR to **1–2 hours (120 users)** with QR.

## Table 2 — Database update and observation resolution time

```python
table2=pd.DataFrame({'Processing time':['2–4 months','1–2 months','1–2 weeks','4–6 days','1–2 days','2–4 hours','1–2 hours','15–30 minutes','2–15 minutes'],'Without QR system':[50,150,100,42,0,0,0,0,0],'With QR system':[0,0,0,0,0,0,22,200,120]})
display(table2)
print('Totals:',table2['Without QR system'].sum(),table2['With QR system'].sum())
```

```python
ax=table2.set_index('Processing time').plot(kind='bar',figsize=(12,7))
ax.set_title('Table 2 — Database Update and Observation Resolution Time',fontweight='bold')
ax.set_xlabel('Processing time'); ax.set_ylabel('Number of users'); ax.legend(title='Condition'); ax.grid(axis='y',alpha=.25)
plt.xticks(rotation=25,ha='right'); plt.tight_layout(); plt.show()
```

**Interpretation:** Without QR, the largest group is **1–2 months (150 users)**. With QR, the largest group is **15–30 minutes (200 users)**, followed by **2–15 minutes (120 users)**.

## Table 3 — Physical/digital observation resolution using the cloud-computing framework
